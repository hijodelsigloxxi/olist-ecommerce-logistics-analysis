# Posibles queries SQL para el análisis logístico

Una vez creadas las capas `core` y `analytics`, el siguiente paso consiste en explotar el modelo mediante consultas SQL. Estas queries servirán para obtener indicadores, detectar patrones logísticos y crear vistas reutilizables para Power BI.

La idea es trabajar principalmente sobre las tablas de la capa `analytics`, especialmente `fact_orders`, `fact_order_items` y sus dimensiones asociadas.

---

## 1. Indicadores generales de rendimiento logístico

**Objetivo:** obtener una visión global del funcionamiento logístico del marketplace.

**Posibles métricas:**

- Total de pedidos.
- Total de pedidos entregados.
- Total de pedidos retrasados.
- Porcentaje de retrasos sobre pedidos entregados.
- Tiempo medio de aprobación.
- Tiempo medio de entrega.
- Días medios de retraso.
- Porcentaje de pedidos entregados antes de la fecha estimada.
- Porcentaje de pedidos entregados dentro del plazo.

**Posibles preguntas:**

- ¿Qué porcentaje de pedidos se entregan tarde?
- ¿Cuál es el tiempo medio total de entrega?
- ¿Cuántos pedidos se entregan antes, dentro o después del plazo estimado?
- ¿Qué proporción de pedidos no llega a completarse?

**Archivo sugerido:**

`sql/analysis/01_general_delivery_kpis.sql`

**Vista posible:**

`analytics.vw_general_order_performance`

## 2. Distribución de los retrasos por días

**Objetivo:** analizar no solo si un pedido se retrasa, sino cuánto se retrasa.

**Posibles rangos:**

- Entregado antes de la fecha estimada.
- Entregado en la fecha estimada.
- Retraso de 1 a 3 días.
- Retraso de 4 a 7 días.
- Retraso de 8 a 15 días.
- Retraso superior a 15 días.

**Posibles métricas:**

- Número de pedidos por rango de retraso.
- Porcentaje de pedidos por rango.
- Review score medio por rango de retraso.
- Valor medio del pedido por rango.
- Coste medio de envío por rango.

**Posibles preguntas:**

- ¿La mayoría de retrasos son leves o graves?
- ¿Qué porcentaje de pedidos retrasados supera una semana de retraso?
- ¿Los retrasos más largos se asocian con peor satisfacción del cliente?

**Archivo sugerido:**

`sql/analysis/02_delay_distribution.sql`

**Vista posible:**

`analytics.vw_delay_distribution`
